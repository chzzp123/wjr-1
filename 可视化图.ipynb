```python
import matplotlib.pyplot as plt
import numpy as np

# 中文设置
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams["axes.unicode_minus"] = False

def main():
    regions = ["华北", "华南", "东北", "西北", "西南", "华东"]
    sales = [2354, 1902, 3524, 2698, 2896, 2563]
    x = np.arange(len(regions))
    width = 0.6

    fig, ax = plt.subplots(figsize=(8, 5))
    bars = ax.bar(x, sales, width=width)

    # 模拟柱子渐变
    for bar in bars:
        height = bar.get_height()
        bar.set_facecolor("none")
        yy = np.linspace(0, height, int(height))
        for y in yy[::8]:
            ax.plot([bar.get_x(), bar.get_x() + width], [y, y], color="#4488dd",
                    alpha=0.2 + (y / height)*0.6, lw=1.2)

    ax.set_xticks(x)
    ax.set_xticklabels(regions)
    ax.set_title("渐变柱形图")
    ax.set_ylabel("销售量")
    plt.tight_layout()
    plt.savefig("chart1_渐变柱形图.png", dpi=150)
    plt.show()

if __name__ == "__main__":
    main()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 中文设置
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams["axes.unicode_minus"] = False

def main():
    regions = ["华北", "华南", "东北", "西北", "西南", "华东"]
    sales = [2354, 1902, 3524, 2698, 2896, 2563]
    mean_val = np.mean(sales)
    x = np.arange(len(regions))

    fig, ax = plt.subplots(figsize=(8, 5))
    ax.bar(x, sales, color="#5599ee")
    ax.axhline(y=mean_val, color="#ff3333", linestyle="--", label=f"均值:{mean_val:.0f}")

    ax.set_xticks(x)
    ax.set_xticklabels(regions)
    ax.set_title("带均值柱形图")
    ax.set_ylabel("销售量")
    ax.legend()
    plt.tight_layout()
    plt.savefig("chart2_带均值柱形图.png", dpi=150)
    plt.show()

if __name__ == "__main__":
    main()
```
