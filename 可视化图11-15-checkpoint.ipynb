```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import make_interp_spline
plt.rcParams["font.family"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

def main():
    data = [
        {"year":2021,"month":"5月","sale":146},
        {"year":2021,"month":"6月","sale":198},
        {"year":2021,"month":"7月","sale":296},
        {"year":2021,"month":"8月","sale":412},
        {"year":2021,"month":"9月","sale":506},
        {"year":2021,"month":"10月","sale":615},
        {"year":2021,"month":"11月","sale":789},
        {"year":2021,"month":"12月","sale":1021},
        {"year":2022,"month":"1月","sale":3782},
        {"year":2022,"month":"2月","sale":3215},
        {"year":2022,"month":"3月","sale":2936},
    ]
    x_labels = [f"{d['year']}{d['month']}" for d in data]
    y = [d["sale"] for d in data]
    x = np.arange(len(x_labels))

    fig, ax = plt.subplots(figsize=(10,5),dpi=120)
    x_smooth = np.linspace(x.min(),x.max(),300)
    spl = make_interp_spline(x, y, k=3)
    y_smooth = spl(x_smooth)

    ax.plot(x_smooth, y_smooth, color="#2E86AB", linewidth=2)
    ax.scatter(x,y,color="#2E86AB",zorder=5,s=25)

    ax.set_xticks(x)
    ax.set_xticklabels(x_labels,rotation=45,ha="right")
    ax.set_title("平滑折线图")
    ax.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig("chart11_平滑折线图.png")
    plt.show()

if __name__ == "__main__":
    main()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Circle, PathPatch
from matplotlib.path import Path

plt.rcParams["font.family"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

def main():
    fig, ax = plt.subplots(figsize=(5,5),dpi=120)
    ax.set_xlim(-1.2,1.2)
    ax.set_ylim(-1.2,1.2)
    ax.set_aspect("equal")
    ax.axis("off")

    outer_circle = Circle((0,0), radius=1, facecolor="none", edgecolor="#3498db", lw=3)
    ax.add_patch(outer_circle)

    percent = 0.65
    h = 2*percent - 1

    theta = np.linspace(np.pi, 2*np.pi, 100)
    x_circle = np.cos(theta)
    y_circle = np.sin(theta)

    x_wave = np.linspace(-1,1,200)
    wave_amp = 0.08
    y_wave = h + wave_amp * np.sin(x_wave*8)

    verts = []
    codes = []
    verts.append((-1, 1))
    codes.append(Path.MOVETO)
    for xi,yi in zip(x_wave,y_wave):
        verts.append((xi,yi))
        codes.append(Path.LINETO)
    for xi,yi in zip(reversed(x_circle),reversed(y_circle)):
        verts.append((xi,yi))
        codes.append(Path.LINETO)
    path = Path(verts,codes)
    wave_patch = PathPatch(path,facecolor="#42a5f5",alpha=0.7,edgecolor="none")
    ax.add_patch(wave_patch)

    ax.text(0,0,f"{percent*100:.0f}%",fontsize=24,ha="center",va="center",color="#0a437a")
    ax.set_title("水球图",pad=20)
    plt.tight_layout()
    plt.savefig("chart15_水球图.png")
    plt.show()

if __name__ == "__main__":
    main()
```
