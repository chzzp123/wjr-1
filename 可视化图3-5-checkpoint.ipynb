```python
import matplotlib.pyplot as plt
import numpy as np

# 中文设置
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams["axes.unicode_minus"] = False

def main():
    goods = ["口红", "面膜", "隔离", "防晒", "精华", "面霜"]
    sales = [653, 523, 648, 856, 714, 785]
    x = np.arange(len(goods))

    fig, ax = plt.subplots(figsize=(8, 5))
    bars = ax.bar(x, sales, width=0.6)

    for bar in bars:
        bar.set_edgecolor("#ee7799")
        bar.set_facecolor("#ffb6c1")

    ax.set_xticks(x)
    ax.set_xticklabels(goods)
    ax.set_title("渐变圆角柱形图")
    ax.set_ylabel("销量")
    plt.tight_layout()
    plt.savefig("chart3_渐变圆角柱形图.png", dpi=150)
    plt.show()

if __name__ == "__main__":
    main()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 中文设置
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams["axes.unicode_minus"] = False

def main():
    goods = ["口红", "面膜", "隔离", "防晒", "精华", "面霜", "眼影", "气垫"]
    sales = [9221, 5102, 6571, 5760, 6321, 8612, 2645, 5321]
    x = np.arange(len(goods))

    fig, ax = plt.subplots(figsize=(9, 5))
    bars = ax.bar(x, sales, color="#66aadd")

    for bar in bars:
        h = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2, h + 120, f"{h}", ha="center", fontsize=9)

    ax.set_xticks(x)
    ax.set_xticklabels(goods)
    ax.set_title("标注柱形图")
    ax.set_ylabel("销量")
    plt.tight_layout()
    plt.savefig("chart4_标注柱形图.png", dpi=150)
    plt.show()

if __name__ == "__main__":
    main()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 中文设置
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams["axes.unicode_minus"] = False

def main():
    quarters = ["2021Q1", "Q2", "Q3", "Q4", "2022Q1", "Q2"]
    sale = [3121, 4086, 4321, 4601, 4936, 4231]
    profit = [1020, 1421, 1502, 1623, 1781, 1432]
    x = np.arange(len(quarters))

    fig, ax = plt.subplots(figsize=(8,5))
    ax.bar(x, sale, label="销售额", color="#4488dd")
    ax.bar(x, profit, bottom=sale, label="利润额", color="#ff9944")

    ax.set_xticks(x)
    ax.set_xticklabels(quarters)
    ax.set_title("层叠柱形图")
    ax.set_ylabel("数值")
    ax.legend()
    plt.tight_layout()
    plt.savefig("chart5_层叠柱形图.png", dpi=150)
    plt.show()

if __name__ == "__main__":
    main()
```
