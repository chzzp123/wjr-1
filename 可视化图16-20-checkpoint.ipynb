```python
import matplotlib.pyplot as plt

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

def main():
    fig, ax = plt.subplots(figsize=(5, 5), subplot_kw={"aspect": "equal"})
    complete = 65
    placeholder = 100 - complete
    data = [complete, placeholder]
    colors = ["#4096ee", "#eeeeee"]
    ax.pie(data, radius=1, colors=colors,
           wedgeprops={"width": 0.3, "edgecolor": "white"},
           startangle=90)
    ax.text(0, 0, f"{complete}%", fontsize=32, ha="center", va="center")
    ax.set_title("本科及以上学历员工占比\n6月30日最新统计", fontsize=12)
    plt.tight_layout()
    plt.savefig("chart16_波浪水球图.png", dpi=150)
    plt.show() 
    plt.close()
    

if __name__ == "__main__":
    main()
```

```python
import matplotlib.pyplot as plt

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

def main():
    fig, ax = plt.subplots(figsize=(6, 6), subplot_kw={"aspect": "equal"})
    labels = [">=50", "[40,50)", "[30,40)", "[20,30)"]
    val = [12.5, 20.8, 29.2, 37.5]
    placeholder = [37.5, 29.2, 20.8, 12.5]
    c_outer = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728"]
    c_inner = ["#e8e8e8"] * 4

    ax.pie(val, radius=1, labels=labels, colors=c_outer,
           wedgeprops={"width": 0.3, "edgecolor": "white"}, startangle=90)
    ax.pie(placeholder, radius=0.7, colors=c_inner,
           wedgeprops={"width": 0.3, "edgecolor": "white"}, startangle=90)

    ax.set_title("玉玦图：员工年龄分布", fontsize=13)
    plt.tight_layout()
    plt.savefig("chart17_玉玦图.png", dpi=150)
    plt.show()
    plt.close()
   

if __name__ == "__main__":
    main()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

def main():
    fig, ax = plt.subplots(figsize=(7, 7), subplot_kw={"aspect": "equal"})
    depts = ["人力部", "行政部", "财务部", "工程部", "采购部", "销售部"]
    count = [130, 226, 238, 293, 326, 451]
    placeholder = [702, 606, 594, 539, 506, 381]

    c_outer = plt.cm.Set3(np.linspace(0, 1, len(depts)))
    c_inner = ["#f0f0f0"] * len(depts)

    ax.pie(count, radius=1, labels=depts, colors=c_outer,
           wedgeprops={"width": 0.25, "edgecolor": "white"}, startangle=90)
    ax.pie(placeholder, radius=0.75, colors=c_inner,
           wedgeprops={"width": 0.25, "edgecolor": "white"}, startangle=90)

    ax.set_title("跑道图：部门人数分布", fontsize=13)
    plt.tight_layout()
    plt.savefig("chart18_跑道图.png", dpi=150)
    plt.show()
    plt.close()


if __name__ == "__main__":
    main()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

def main():
    depts = ["销售部", "采购部", "工程部", "财务部", "行政部", "人力部"]
    ratio = np.array([29.2, 22.7, 17.5, 13.6, 10.3, 6.7])
    theta = np.linspace(0, 2 * np.pi, len(depts), endpoint=False)
    width = 2 * np.pi / len(depts)

    fig, ax = plt.subplots(figsize=(7, 7), subplot_kw={"projection": "polar"})
    ax.bar(theta, ratio, width=width, bottom=0.0, color=plt.cm.tab10(np.arange(len(depts))))
    ax.set_xticks(theta)
    ax.set_xticklabels(depts)
    ax.set_yticks([])
    ax.set_title("南丁格尔圆饼图 部门人数占比", pad=20)

    plt.tight_layout()
    plt.savefig("chart19_南丁格尔圆饼图.png", dpi=150)
    plt.show()
    plt.close()

if __name__ == "__main__":
    main()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

def main():
    age_labels = ["[20,30)", "[30,40)", "[40,50)", ">=50"]
    ratio = np.array([37.5, 29.2, 20.8, 12.5])
    theta = np.linspace(0, 2 * np.pi, len(age_labels), endpoint=False)
    width = 2 * np.pi / len(age_labels)

    fig, ax = plt.subplots(figsize=(7, 7), subplot_kw={"projection": "polar"})
    ax.bar(theta, ratio, width=width, bottom=20, color=plt.cm.Set2(np.arange(len(age_labels))))
    ax.set_xticks(theta)
    ax.set_xticklabels(age_labels)
    ax.set_yticks([])
    ax.set_title("南丁格尔圆环图 年龄占比", pad=20)

    plt.tight_layout()
    plt.savefig("chart20_南丁格尔圆环图.png", dpi=150)
    plt.show()
    plt.close()
    
if __name__ == "__main__":
    main()
```
